```python

#Lecture 1 Displaying the image using a window

import cv2
img = cv2.imread(r'/home/ayush-ubantu/Downloads/Jamia/Lab/Sem3/MS36/Lec1/mickey_colored.png')
cv2.imshow("Image",img)

cv2.waitKey(0)
cv2.destroyAllWindows()

#resizing the image and displaying it
resize = cv2.resize(img,(100,100))
cv2.imshow("Image",resize)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# ROI = image[row_start:row_end, column_start:column_end]
roi = img[0:230, 60:460]
cv2.imshow("ROI", roi)
print(roi.shape)   # (230, 400, 3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
h, w = img.shape[:2]
ratio = 800 / w
dim = (800, int(h * ratio))
resize_aspect = cv2.resize(img, dim)
cv2.imshow("Resized Image", resize_aspect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
horizontal = cv2.flip(img, 1)   # flip around y-axis (mirror left-right)
vertical   = cv2.flip(img, 0)   # flip around x-axis (upside down)
both       = cv2.flip(img, -1)  # flip both axes

cv2.imshow("Horizontal", horizontal)
cv2.imshow("Vertical", vertical)
cv2.imshow("Both", both)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
path = "mickey_colored.png"
src = cv2.imread(path)

window_name = "Image"

# Using cv2.ROTATE_90_CLOCKWISE - rotate by 90 degrees clockwise
image = cv2.rotate(src, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow(window_name, image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread('mickey_colored.png')
(rows, cols) = img.shape[:2]

# Rotation matrix for a 45-degree rotation about the image center, no scaling
M = cv2.getRotationMatrix2D((cols / 2, rows / 2), 45, 1)
print(M)

res = cv2.warpAffine(img, M, (cols, rows))
cv2.imshow("Rotated image", res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
output = img.copy()

# pt1 = top-left corner, pt2 = bottom-right corner (any two opposite corners work)
cv2.rectangle(output, (100, 0), (420, 260), (255, 0, 0), 2)

cv2.imshow("Rectangle", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
output=cv2.circle(output, (260, 130), 100, (255, 0, 0), 2)
```

```python
output = img.copy()

# putText(image, text, bottom_left_origin, font, font_scale, color(BGR), thickness)
cv2.putText(output, "OpenCV Demo", (60, 490), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)

cv2.imshow("Text", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
