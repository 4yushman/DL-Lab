```python
#Binary cross entropy loss function
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8  # Small value to avoid log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)  # Clamp predictions to avoid log(0)
    return - (target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
#Forward pass
z = w*x + b #weighted sum which is linear part
y_pred = torch.sigmoid(z) #Predicted Probability

#Compute the binary cross-entropy loss
loss = binary_cross_entropy_loss(y_pred, y)
loss
```

```python
#Doing Backpropagation to compute gradients of loss w.r.t. parameters w and b
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```
