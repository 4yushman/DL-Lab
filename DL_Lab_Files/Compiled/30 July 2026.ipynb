```python
#resizing the image and selecting a ROI (Region of interest)

w,h = img.shape[:2]

#Problem with roi is that aspect ratio of image is not maintained.
#so we need to do some etra work in order to maintain a proper aspect ratio

h, w = img.shape[:2]          # correct unpacking

ratio = 800 / w               # scale factor based on target width
dim = (800, int(h * ratio))   # resize wants (width, height)

resize_aspect = cv2.resize(img, dim)
cv2.imshow("res", resize_aspect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#FLipping an image
horizontal = cv2.flip(img,1)
cv2.imshow("horizontal_flip",horizontal)
cv2.waitKey(0)
cv2.destroyAllWindows()

#vertical
vertical = cv2.flip(img,0)
cv2.imshow("vertical _flip",vertical )
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#rotating iages
rotated_img = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("90_clockwise",rotated_img )
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#creating a rectangle on top of an image

output = img.copy()
rectangle = cv2.rectangle(output,(10,40),(20,40),(255,0,0),2)

rotated_img = cv2.rotate(rectangle,cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("90_clockwise",rotated_img )
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#scalar(0 dim)
x0 =torch.tensor(1)

#1d 
x1 = torch.tensor([1])

#2d
x2 = torch.tensor([[1]])
#and so on

print(x0)
print(x1)
print(x2)

#Example of a 4d tensor
#Num_of_images X Num_of_channels X Height X Width
```

```python
#Multiple ways to create tensors
#1) directly from data
tensor_1 = torch.tensor([[1,2],[3,4]])

#2) data from numpy array
import numpy as np
tensor_2 = torch.tensor(np.array([[1.0, 2.0], [2.0, 3.0]])) 

# 3) from another tensor
tensor_3 = torch.ones_like(tensor_2)
tensor_4 = torch.rand_like(tensor_2)
#here the new tensor retaons properties(shape, datatype) of argument tensor, unless explicitly overwritten


print(tensor_1)

print(tensor_2)
print(tensor_3)
print(tensor_4)
```

```python
#another way of doing it
shape = (2,3,)

rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
print(rand_tensor)
print(ones_tensor)
print(zeros_tensor)
```
